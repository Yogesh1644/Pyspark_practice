```
from pyspark.sql.types import*
data=[(1,'Riya'),(2,'Sonu')]
schema=StructType([StructField("id", dataType=IntegerType(),nullable=True),
            StructField("name", dataType=StringType(),nullable=True)
            ])
df=spark.createDataFrame(data=data, schema=schema)
df.show()
df.printSchema()
```
